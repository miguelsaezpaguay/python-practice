```python
import numpy as np
```

```python
data = np.loadtxt(fname='data/inflammation-03.csv', delimiter= ',')

max_inf_0 = np.max(data, axis=0)[0]  # Maximum inflammation value on day zero
max_inf_20 = np.max(data, axis=0)[20] # Maximum inflammation value on day twenty

if max_inf_0 == 0 and max_inf_20 == 20:
    print('Suspicious-looking maxima!')

elif np.sum(np.min(data, axis=0)) == 0:
    print('Minima add up to zero')

else:
    print('Seems OK')
```
